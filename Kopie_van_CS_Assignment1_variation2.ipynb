````python
# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

DATA_DIR = "/content/drive/MyDrive/Prinsjesdag/Preprocessed"

import os, re, glob
from collections import Counter, defaultdict
import pandas as pd
import numpy as np

# Stopwoordenlijst (zoals in de opdracht)
stopwords = [
    "aan", "af", "al", "alles", "als", "altijd", "andere", "ben", "bij", "daar",
    "dan", "dat", "de", "der", "deze", "die", "dit", "doch", "doen", "door",
    "dus", "een", "eens", "en", "er", "ge", "geen", "geweest", "haar", "had",
    "heb", "hebben", "heeft", "hem", "het", "hier", "hij", "hoe", "hun", "iemand",
    "iets", "ik", "in", "is", "ja", "je", "kan", "kon", "kunnen", "maar", "me",
    "meer", "men", "met", "mij", "mijn", "moet", "na", "naar", "niet", "niets",
    "nog", "nu", "of", "om", "omdat", "ons", "onze", "ook", "op", "over", "reeds",
    "te", "tegen", "toch", "toen", "tot", "u", "uit", "uw", "van", "veel",
    "voor", "want", "waren", "was", "wat", "we", "wel", "werd", "wezen", "wie",
    "wij", "wil", "worden", "zal", "ze", "zei", "zelf", "zich", "zij", "zijn",
    "zo", "zonder", "zou"
]
STOP = set(stopwords)

def year_from_filename(path):
    name = os.path.basename(path)
    m = re.search(r'(20\d{2})', name)
    return int(m.group(1)) if m else None

def read_words(path):
    with open(path, 'r', encoding='utf-8') as f:
        text = f.read().lower()
    return text.split()

files = sorted(glob.glob(os.path.join(DATA_DIR, "*.txt")))
data = []

for fp in files:
    y = year_from_filename(fp)
    if y is None:
        continue
    # Alleen speeches van Willem-Alexander (2013–2025)
    if 2013 <= y <= 2025:
        words = read_words(fp)
        data.append({"year": y, "path": fp, "words": words})

# Controle of het werkt
years_loaded = sorted([d["year"] for d in data])
print("Ingeladen jaren:", years_loaded)
print("Aantal bestanden:", len(data))

import os

for root, dirs, files in os.walk("/content/drive/MyDrive"):
    if files:
        print("MAP:", root)
        for f in files:
            if f.endswith(".txt"):
                print("   →", f)
rows = []
klimaat_counts = {}
duurzaam_counts = {}
stopword_pct = {}
unique_counts = {}
total_counts = {}

suffix_targets = ["heid", "ing", "schap"]
suffix_totals = Counter()
heid_counter_global = Counter()

longest_word_global = ""

for item in data:
    y = item["year"]
    words = item["words"]
    total = len(words)
    uniques = len(set(words))
    total_counts[y] = total
    unique_counts[y] = uniques

    # "klimaat" als substring in een woord
    klimaat_counts[y] = sum(1 for w in words if "klimaat" in w)

    # woorden die beginnen met "duurzaam"
    duurzaam_counts[y] = sum(1 for w in words if w.startswith("duurzaam"))

    # percentage stopwoorden
    sw = sum(1 for w in words if w in STOP)
    stopword_pct[y] = (sw / total) * 100 if total > 0 else 0

    # suffix-tellingen
    for w in words:
        for suf in suffix_targets:
            if w.endswith(suf):
                suffix_totals[suf] += 1
        if w.endswith("heid"):
            heid_counter_global[w] += 1

    # langste woord
    candidate = max(words, key=len)
    if len(candidate) > len(longest_word_global):
        longest_word_global = candidate

# DataFrame met resultaten per jaar
df = pd.DataFrame({
    "year": sorted(total_counts.keys())
})
df["total_words"] = df["year"].map(total_counts)
df["unique_words"] = df["year"].map(unique_counts)
df["klimaat_count"] = df["year"].map(klimaat_counts)
df["duurzaam*_count"] = df["year"].map(duurzaam_counts)
df["stopword_pct"] = df["year"].map(stopword_pct)

df

````
